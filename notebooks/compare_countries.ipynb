```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import f_oneway

# load the cleaned CSVs
df_benin = pd.read_csv("../data/benin_clean.csv")
df_sierra = pd.read_csv("../data/sierraleone_clean.csv")
df_togo = pd.read_csv("../data/togo_clean.csv")


df_benin['country'] = 'Benin'
df_sierra['country'] = 'Sierra Leone'
df_togo['country'] = 'Togo'

df_all = pd.concat([df_benin, df_sierra, df_togo], ignore_index=True)


plt.figure(figsize=(12,6))
sns.boxplot(data=df_all, x='country', y='GHI')
plt.title("GHI Distribution by Country")
plt.show()

plt.figure(figsize=(12,6))
sns.boxplot(data=df_all, x='country', y='DNI')
plt.title("DNI Distribution by Country")
plt.show()

plt.figure(figsize=(12,6))
sns.boxplot(data=df_all, x='country', y='DHI')
plt.title("DHI Distribution by Country")
plt.show()


def summary_stats(df, col):
    return pd.Series({
        'mean': df[col].mean(),
        'median': df[col].median(),
        'std': df[col].std()
    })

summary = pd.concat([
    summary_stats(df_benin, 'GHI'),
    summary_stats(df_sierra, 'GHI'),
    summary_stats(df_togo, 'GHI')
], axis=1)

summary.columns = ['Benin', 'Sierra Leone', 'Togo']
summary


fstat, pval = f_oneway(
    df_benin['GHI'].dropna(),
    df_sierra['GHI'].dropna(),
    df_togo['GHI'].dropna()
)

print("ANOVA F-statistic:", fstat)
print("p-value:", pval)


```

### Key Observations
1. Benin shows the highest average GHI.
2. Sierra Leone data has the largest variability.
3. Togo’s median GHI is close to Benin’s but with fewer outliers.

### Recommendations
1. Prioritize Benin for pilot solar installations due to stable high GHI.
2. Further clean and analyze Sierra Leone’s data before investment decisions.
